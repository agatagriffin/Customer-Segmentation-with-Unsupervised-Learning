#FEATURE ENGINEERING

```python
# Copy my_data and create data_convert
data_convert = my_data.copy()
```

```python
# define currency_to_float funciton
def currency_to_float(currency_str):
    """Converts a currency string to a float."""
    if isinstance(currency_str, str):
        # Remove '$', ',', and '()'
        currency_str = currency_str.replace('$', '').replace(',', '').replace('(', '-').replace(')', '')
    # Convert to numeric, coercing errors
    return pd.to_numeric(currency_str, errors='coerce')
```

```python
# Create Total Revenue and Unit Cost Features in numerical values
data_convert['Total Revenue'] = data_convert['Total Revenue'].apply(currency_to_float)
data_convert['Unit Cost'] = data_convert['Unit Cost'].apply(currency_to_float)
```

```python
# Create Frequency feature
data_convert['Frequency'] = data_convert.groupby('Customer ID')['Order ID'].transform('count')

print(data_convert)
```

```python
# Show table dimensions
print("Table dimensions (rows, columns):", data_convert.shape)
```

```python
# Create Recency Feature

data_convert['Delivery_Date'] = pd.to_datetime(data_convert['Delivery_Date'], format='%d%b%Y', errors='coerce')
data_convert['Recency'] = (pd.to_datetime('today') - data_convert['Delivery_Date']).dt.days

display(data_convert[['Delivery_Date', 'Recency']].head())
```

```python
# Create CLV Feature
data_convert = data_convert.rename(columns={'Total Revenue': 'CLV'})

display(data_convert[['CLV']].head())

```

```python
# Create Avg Unit Cost Feature
data_convert = data_convert.rename(columns={'Unit Cost': 'Avg Unit Cost'})
```

```python
# Create DOB feature in numerical format

data_convert['DOB'] = pd.to_datetime(data_convert['Customer_BirthDate'], format='%d%b%Y', errors='coerce')

data_convert ['Age'] = (pd.to_datetime('today') - data_convert['DOB']).dt.days // 365
```

# Clustering Analysis - selection of k for k-means

1. Elbow method is used to identify optimal k for k-means model input

```python
# Import Kmeans function.
from sklearn.cluster import KMeans

# Elbow method: try different k.
X = data_final_standardized
wcss=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i, init='k-means++',random_state=0, n_init = 10)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.plot(range(1,11),wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

Whilst there is no clear-cut answer as to the number of clusters identified by the Elbow method, some observations in support of the level of k-means can be made. The rate of decrease slows at k=2 as the line turns from near linearly decreasing to a curved shape.  Similarly, a slight kink in the curve is observed at n=5 and n=6, however the rate of WCSS decrease is not constant. Hence, either of the points could be indicative of an optimal number of clusters in that intra-cluster distances could not be meaningfully reduced further for a given data set. Further analysis is needed to determine an optimal k for k-means calculation, with k=2, k=5 and k=6 as possible options. Further analysis is needed to determin an optimal k for k-means calculation.

2. Silhouette method is used to identify optimal k for k-means model input

```python
# Import necessary library.
from sklearn.metrics import silhouette_samples, silhouette_score
range_n_clusters = [2, 3, 4, 5, 6, 8, 10]
```

```python
for n_clusters in range_n_clusters:

    clusterer = KMeans(n_clusters=n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(X)

    silhouette_avg = silhouette_score(X, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(X, cluster_labels)
```

The model generated the highest Silhouette coefficient of 0.2668, indicating an optimal number of clusters of k=5. This corresponded to the kink in the curve observed on the Elbow Method graph. The second highest silhouette coefficient of 0.2621 was at k=2, similar to the drop in the rate of decrease of inertia on the Elbow Method chart.  The silhouette coefficient keeps decreasing from k=5, precluding k=6 as an optimal choice, guiding a plausible range of coeffcient value between k=2 and k=5, with k=5 being a more efficient silhouette coefficient as its value is highest. This is further supported by the Silhoutte graphs where, although each of the tested possible k-means had negative coefficient values, there were relatively fewer of them overall, observed for k=5.

Silhoutte method - graphs

```python
import matplotlib.cm as cm # Import colormap


for n_clusters in range_n_clusters:
    # Create a subplot with 1 row and 2 columns.
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot.
    # The silhouette coefficient can range from -1, 1 but in this example all
    # lie within [-0.1, 1].
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(X) + (n_clusters + 1) * 10])

    # Initialize the clusterer with n_clusters value and a random generator
    # seed of 10 for reproducibility.
    clusterer = KMeans(n_clusters=n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(X)

    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(X, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette score/coefficient for each sample
    sample_silhouette_values = silhouette_samples(X, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them.
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle.
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot.
        y_lower = y_upper + 10  # 10 for the 0 samples.

    ax1.set_title("The silhouette plot for the various clusters.")
    ax1.set_xlabel("The silhouette coefficient values")
    ax1.set_ylabel("Cluster label")

    ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

    ax1.set_yticks([])  # Clear the yaxis labels / ticks.
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

    # 2nd Plot showing the actual clusters formed.
    colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
    ax2.scatter(X[:, 0], X[:, 1], marker='.', s=30, lw=0, alpha=0.7,
                c=colors, edgecolor='k')

    # Labeling the clusters.
    centers = clusterer.cluster_centers_
    # Draw white circles at cluster centers.
    ax2.scatter(centers[:, 0], centers[:, 1], marker='o',
                c="white", alpha=1, s=200, edgecolor='k')

    for i, c in enumerate(centers):
        ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1,
                    s=50, edgecolor='k')

    ax2.set_title("The visualization of the clustered data.")
    ax2.set_xlabel("Feature space for the 1st feature")
    ax2.set_ylabel("Feature space for the 2nd feature")

    plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                  "with n_clusters = %d" % n_clusters),
                 fontsize=14, fontweight='bold')

plt.show()
```

Based on the Silhouette analysis, it is difficult to derive meaningful Silhouette coefficient as n=3..n=6, all have negative scores which weakens their reliability. k=5 again emergest as a plausible candidate as although there are negative values of the Silhouette tte coefficent observed for 3 clusters the lines are rather thin meaning few observations fall in this category.

3. Hierarchical clustering is used to identify optimal k for k-means model input

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import StandardScaler # Import StandardScaler

# Sample a smaller subset of data_final for hierarchical clustering
sample_size = 30000
Small_df = data_final.sample(n=sample_size, random_state=42).reset_index(drop=True) # Added random_state for reproducibility

# Standardize the sampled data
# Fit a new scaler on the smaller sample
sample_scaler = StandardScaler()
Small_df_standardized = sample_scaler.fit_transform(Small_df)


# Perform Hierarchical Agglomerative Clustering (on the standardized sample)
# Choose a number of clusters (e.g., 3 as in the commented code, or you can adjust)
n_clusters_hierarchical = 3
agglo_cluster = AgglomerativeClustering(n_clusters=n_clusters_hierarchical, metric='euclidean', linkage='average')
# Fit and predict clusters on the standardized sample
agglo_cluster_labels = agglo_cluster.fit_predict(Small_df_standardized)

# Add cluster labels to the sampled DataFrame
Small_df['cluster'] = agglo_cluster_labels

# Create the linkage matrix for the dendrogram
# Use the standardized sample data for linkage calculation
Z = linkage(Small_df_standardized, method='average')

# Plotting the dendrogram
plt.figure(figsize=(15, 10))
# Truncate the dendrogram to show the last 'p' merged clusters
dendrogram(Z, truncate_mode='lastp', p=10, leaf_rotation=90, leaf_font_size=12) # Adjusted p for a reasonable number of leaves
plt.title('Hierarchical Clustering Dendrogram (Truncated)')
plt.xlabel(f'Sample Index or (Size of Cluster) - Showing last {10} merges')
plt.ylabel('Distance')
plt.show()

```

Hierarchical clustering run on a sample of 30k observations shows that there 2 or 4 clusters on the dendogram. This method, whilst deterministic, has its limiations as it requires large computational resources for larger data sets, and cannot be conclusive for this analysis given the sample of 30,000 observation was used.

The subsequent analysis will be performed with k=5 as this number of clusters was identified as optimal by the Silhouette method.
